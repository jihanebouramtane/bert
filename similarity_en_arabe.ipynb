# **similarity en Arab :**

```python
from sentence_transformers import SentenceTransformer
from sklearn.metrics.pairwise import cosine_similarity
import pandas as pd

# Liste de phrases en arabe
sentences = [
    "أنا أحب البرمجة",
    "البرمجة ممتعة ومفيدة",
    "أذهب إلى السوق كل يوم",
    "الطقس جميل اليوم في المدينة",
    "أعمل على مشروع جديد باستخدام بايثون",
    "أحب مشاهدة الأفلام في وقت الفراغ"
]
```

Ce code utilise le modèle multilingue paraphrase pour encoder une liste de phrases en arabe.

```python
# Chargement du modèle multilingue
model = SentenceTransformer('sentence-transformers/paraphrase-multilingual-MiniLM-L12-v2')
```

```python
# Encodage des phrases
embeddings = model.encode(sentences)
```

```python
# Calcul de la similarité et organisation dans un tableau
results = []
threshold = 0.7  # seuil à partir duquel deux phrases sont considérées comme similaires

for i in range(len(sentences)):
    for j in range(i + 1, len(sentences)):  # éviter la comparaison avec soi-même
        score = cosine_similarity([embeddings[i]], [embeddings[j]])[0][0]
        results.append({
            "Phrase 1": sentences[i],
            "Phrase 2": sentences[j],
            "Similarité (%)": round(score * 100, 2),
            "Similaire ?": " Oui" if score >= threshold else "Non"
        })

df = pd.DataFrame(results)
df
```
