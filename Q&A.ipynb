# Q&A with finetuned BERT



```python
import torch
from transformers import BertForQuestionAnswering, BertTokenizer
```


 nous téléchargeons et chargeons le modèle. Nous utilisons le modèle bert-large-uncased-whole-word-masking-finetuned-squad, qui a été fine tuné sur le jeu de données SQuAD (Stanford Question Answering Dataset).

.


```python
model = BertForQuestionAnswering.from_pretrained('bert-large-uncased-whole-word-masking-finetuned-squad')
```

 nous téléchargeons et chargeons le tokenizer


```python
tokenizer = BertTokenizer.from_pretrained('bert-large-uncased-whole-word-masking-finetuned-squad')
```



## Prétraitement de l’entrée

Tout d'abord, nous définissons les entrées pour BERT, à savoir la question et le paragraphe de texte.


```python
question = "What is the immune system?"
paragraph = "The immune system is a system of many biological structures and processes within an organism that protects against disease. To function properly, an immune system must detect a wide variety of agents, known as pathogens, from viruses to parasitic worms, and distinguish them from the organism's own healthy tissue."
```

```python
question = '[CLS] ' + question + '[SEP]'
paragraph = paragraph + '[SEP]'
```

```python
question_tokens = tokenizer.tokenize(question)
paragraph_tokens = tokenizer.tokenize(paragraph)
```



Combinez tokens de la question et du paragraphe, puis convertissez-les en input_ids .

```python
tokens = question_tokens + paragraph_tokens
input_ids = tokenizer.convert_tokens_to_ids(tokens)
```



 nous définissons les segment_ids. Les segment_ids seront à 0 pour tous les jetons de la question, et à 1 pour tous les jetons du paragraphe.




```python
segment_ids = [0] * len(question_tokens)
segment_ids += [1] * len(paragraph_tokens)
```


Nous convertissons maintenant les input_ids et les segment_ids en tensors.

```python
input_ids = torch.tensor([input_ids])
segment_ids = torch.tensor([segment_ids])
```





## Obtention de la réponse

Nous fournissons les input_ids et les segment_ids au modèle, qui retourne un score de début (start score) et un score de fin (end score) pour chacun des jetons.

```python
start_scores, end_scores = model(input_ids, token_type_ids = segment_ids)
```

```python
start_index = torch.argmax(start_scores)
end_index = torch.argmax(end_scores)
```

C'est terminé ! Nous affichons maintenant la portion de texte entre l'indice de début et l'indice de fin comme notre réponse:

```python
print(' '.join(tokens[start_index:end_index+1]))
```
